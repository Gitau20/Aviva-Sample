#  **FASTER R-CNN AND YOLOv3**
Faster R-CNN is a two-stage algorithm for object detection. First, it uses two networks: a network to detect the location of an object and a network to determine what the object is. The specific flow is that first, the data output from the network to detect the location of the object is input to the network to determine what the object is. Here, the network that specifically finds the location of the object is called the Region Proposal Network (RPN).

# Bounding Box
The rectangle that surrounds an object in the ground truth data is called a Bounding Box (BBox). Using this, we can accurately determine the object's position.
# Object location detection network RPN
# Anchor Box
First, create rectangles of various shapes in the image and search for which rectangles contain objects. These rectangles are called anchor boxes. Object detection networks such as Faster R-CNN create many anchor boxes to distinguish between overlapping backgrounds and objects. The positions and sizes of the anchor boxes are learned to minimize loss. In the field, trial and error of the number and shapes of anchor boxes is the key to success.

Network Structure

- Generate feature maps by running them through a large CNN model (e.g. VGG16)
- Create a network that takes a feature map as input and outputs whether it is an object or background
- Create a network that uses the same feature map to output the overlap
- Train each network

This results in a structure with one input and two outputs.

# Object Classification Network
# Network Structure
- Reusing feature maps generated by RPN
- The feature map is converted to a fixed size using a method called ROI Pooling. This implementation makes it independent of the size of the input image.
- Input the output of ROI Pooling to a classification network
- The same output is used to feed a network that corrects the misalignment of the rectangles.
- Train each network

In practical applications, this method can perform classification with high accuracy, and is particularly useful in scenes where multiple objects exist side by side.

# ROI Pooling
ROI Pooling is a process that corrects objects to a uniform size even if the objects have different sized BBoxes, thereby reducing the error at the pixel level.

Faster R-CNN is completed by combining a network for detecting object positions (RPN) and a network for classifying objects.

````python
# Import necessary libraries
import pandas as pd
from sklearn.preprocessing import LabelEncoder
````

````python
import pandas as pd

# FILE PATHS
input_annotation = '/content/keras-yolo3/annotation.txt'  # annotations file
output_train_txt = 'train.txt'       # YOLO-compatible train file
output_classes_txt = 'classes.txt'   # List of class names (in order)

# STEP 1: Load and Parse annotation.txt
# Format: ./path/image.jpg,xmin,ymin,xmax,ymax,class_name
df = pd.read_csv(input_annotation, header=None)
df.columns = ['image_path', 'xmin', 'ymin', 'xmax', 'ymax', 'class_name']
df.head()
````

````python
# STEP 2: Map class names to numeric IDs
class_names = sorted(df['class_name'].unique())
class_to_id = {name: idx for idx, name in enumerate(class_names)}

# Save to classes.txt
with open(output_classes_txt, 'w') as f:
    for class_name in class_names:
        f.write(class_name + '\n')

# STEP 3: Write YOLO-style train.txt
with open(output_train_txt, 'w') as f:
    for _, row in df.iterrows():
        img_path = row['image_path']
        xmin = int(row['xmin'])
        ymin = int(row['ymin'])
        xmax = int(row['xmax'])
        ymax = int(row['ymax'])
        class_id = class_to_id[row['class_name']]
        f.write(f"{img_path} {xmin},{ymin},{xmax},{ymax},{class_id}\n")

print(f" Done! Saved:\n- train.txt ({output_train_txt})\n- classes.txt ({output_classes_txt})")
````

## Summary Table: YOLOv3 Paper Concepts and Code Mapping

| YOLOv3 Paper Concept         | Description                                                                 | Code Location                                                |
|-----------------------------|-----------------------------------------------------------------------------|--------------------------------------------------------------|
| **1. Backbone: Darknet-53** | Deep CNN with residual connections for feature extraction.                 | `yolo_body()` in `yolo3/model.py`                            |
| **2. Multi-scale predictions** | 3 output scales: 13×13, 26×26, 52×52 for large, medium, small objects.     | `yolo_body()` returns 3 outputs; 3 `y_true` inputs used.     |
| **3. Anchor boxes**         | Predefined boxes guiding prediction shapes.                                | `get_anchors()` and `preprocess_true_boxes()`                |
| **4. Bounding box encoding**| Uses sigmoid and exponential for (tx, ty, tw, th) relative to anchors.      | `preprocess_true_boxes()` and `yolo_loss()` in `model.py`   |
| **5. Objectness score**     | Binary cross-entropy to estimate if an object exists.                      | Calculated in `yolo_loss()`                                  |
| **6. Class prediction**     | Sigmoid + binary cross-entropy for multilabel classification.              | `yolo_loss()`                                                |
| **7. Ignore threshold (0.5)**| Low-IoU predictions ignored in training.                                   | `ignore_thresh=0.5` passed to `yolo_loss()` Lambda           |
| **8. Transfer learning**    | Pretrained weights loaded; layers frozen during initial training.          | `create_model()` and `freeze_body` in `train.py`            |
| **9. Training pipeline**    | Split into two phases; learning rate schedule and early stopping used.     | `_main()` in `train.py`                                     |
| **10. Data augmentation**   | Random scaling, flipping, HSV color adjustment during training.            | `get_random_data()` in `yolo3/utils.py`                      |
| **11. Loss function**       | Sum of bounding box loss, objectness loss, class loss.                     | `yolo_loss()` in `yolo3/model.py`                            |
| **12. Tiny YOLO variant**   | Smaller network with 2 scales for mobile or embedded use.                  | `tiny_yolo_body()` and `create_tiny_model()`                |

## Code Structure Overview

### 1. `train.py`
- Loads annotations and anchors
- Builds model (`create_model()` or `create_tiny_model()`)
- Freezes backbone for first training phase
- Unfreezes for fine-tuning
- Uses:
  - `ReduceLROnPlateau`
  - `EarlyStopping`
  - `ModelCheckpoint`

### 2. `yolo3/model.py`
- `yolo_body()`: Builds full YOLOv3 model with 3 outputs
- `tiny_yolo_body()`: Smaller version for edge devices
- `yolo_loss()`: Custom loss function combining bounding box, objectness, and class loss
- `preprocess_true_boxes()`: Formats ground truth into YOLO-friendly format

### 3. `yolo3/utils.py`
- `get_random_data()`: Applies data augmentation and loads images/labels for training
